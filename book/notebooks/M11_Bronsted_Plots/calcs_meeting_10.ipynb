The code below will plot the $\log{k_{obs}}$ for a reaction vs the $pK_a$ of different acids or bases used. The slope will be a qualitative measure of the amount of proton transfer in the rate-determining step. If the slope is $1$, then the proton transfer is before the rds (likely specific catalysis), if the slope is an intermediate value, the proton transfer is during the rds (general catalysis). lower values imply lesser extent of proton transfer in the transition state and vice versa. If the slope is zero then proton transfer in not involved in the rate law. The transfer occurs after the rds or the mechanism does not rely on proton transfer at all.

## Acid Catalysis

In the code below, I demonstrate the Brønsted plot for acid calaysis. I obtained a set of data for the $k_{obs}$ for the mutarotation of glucose in the presence of different acids whith different $K_a$ values. The data set is a set of values for $k_{obs}$ and a corresponding set of values for the $K_a$ of each acid used.

The data came from a paper published almost 100 years ago by Brønsted himself. It is fascinating to read. Can you discern their method? 

 Brønsted, J.N.; Guggenheim, E.A., "Contribution to the Theory of Acid and Basic Catalysis. The Mutarotation of Glucose." *J. Am. Chem. Soc.*, **1927**, *49*, 2554-2584. [https://doi.org/10.1021/ja01409a031](https://doi.org/10.1021/ja01409a031). Subscription access.

```python
from matplotlib import pyplot as plt
import numpy as np
from scipy.stats import linregress

# Data from Contribution to the Theory of Acid and Basic Catalysis. The Mutarotation of Glucose
# J. N. Brönsted and E. A. Guggenheim
# Journal of the American Chemical Society 1927 49 (10), 2554-2584
# DOI: 10.1021/ja01409a031

# Data entered as a list of x,y pairs. 
data = [[0.10, 2.0], 
        [0.14, 2.1],
        [0.18, 2.4],
        [0.5,  2.8],
        [1.4,  5.6],
        [2.1,  4.6],
        [4.3,  5.7],
        [15,   6.8]]

data = np.array(data)            # convert list of pairs into a n X 2 array
K_a_array = data[:,0] * 1E-4     # take the first item in each pair (first "column" of array)
k_acid_array = data[:,1] * 1E-3  # take the second item in each pair (second "column" of array)
# Note: The factors 1E-4 and 1E-3 come from the data table in the paper.

x = -np.log10(K_a_array)         # pKa
y = np.log10(k_acid_array)       # log k_obs

# Most of the code below was generated in the google AI summary when I searched 
#  line fitting with python. Programmers will be replaced but no AI chatbot
#  can measure amounts and set up a distilation column. You are safe - for now.

# perform line fit and collect results in variables
slope, intercept, r_value, p_value, std_err = linregress(x, y)

# Generate points for the fitted line
x_line = np.linspace(min(x), max(x), 100)
y_line = slope * x_line + intercept

# Plot the data and the fitted line
plt.scatter(x, y)
plt.plot(x_line, y_line, color='red')

plt.xlabel("pKa")
plt.ylabel('log k')

plt.show()

print(f"slope = {slope:0.3f} +/- {std_err:0.3f}")
print(f"rsq = {r_value**2:0.4f}")

#plt.show()
```

The slope of the Brønsted equation for acid catalysis, $-\alpha$, describes the extend of proton transfer from the acid in the transition state. The Brønsted equation is...

\begin{equation*}\log{k_{obs}} = -\alpha \cdot pK_a + C\end{equation*} 

If we plot $\log{k_{obs}}$ vs $pK_a$ we will obtain the slope, $-\alpha$.

## Base Catalysis

Table VII in the paper also contains the values for $k_{obs}$ in the presence of the basic form of the acid. We can plot the base-dependant values for $\log{k_{obs}}$ vs $pK_a$ of the corresponding acid. In base catalysis, the Brønsted equation is written as... 

\begin{equation*}\log{k_{obs}} = \beta \cdot pK_a + C\end{equation*} 

and the slope, $\beta$ describes the extent of proton transfer from the reactant to the base in the rds.

```python
# THIS CODE IS COPIED FROM ABOVE ##############################


# Data from Contribution to the Theory of Acid and Basic Catalysis. The Mutarotation of Glucose
# J. N. Brönsted and E. A. Guggenheim
# Journal of the American Chemical Society 1927 49 (10), 2554-2584
# DOI: 10.1021/ja01409a031

## CHOOSE DATA      ##################################################

# We can select between the two data sets with a simple "if" structure

data_set = "acid"    # select acid or base

# Data entered as a list of x,y pairs. 
if data_set == "acid":
    data = [[0.10, 2.0], 
            [0.14, 2.1],
            [0.18, 2.4],
            [0.5,  2.8],
            [1.4,  5.6],
            [2.1,  4.6],
            [4.3,  5.7],
            [15,   6.8]]
    x_ticks = [3,4,5]
    y_ticks = [-2.8,-2.4,-2.0]
    r_ticks = [-0.2,0,0.2]
else:
    data = [[0.10,31.4], 
            [0.14,28.1], 
            [0.18,26.5], 
            [0.5, 20.0], 
            [0.6, 15.2], 
            [1.3, 12.2], 
            [1.4, 13.7], 
            [2.1, 16.5], 
            [4.3, 10.8], 
            [10,  4.6 ], 
            [13,  6.4 ], 
            [15,  5.4 ], 
            [35,  3.8 ]] 
    x_ticks = [2,3,4,5]
    y_ticks = [-2.5,-2.0,-1.5]
    r_ticks = [-0.2,0,0.2]

import numpy as np              # I had to import numpy here because I am using it before the imports later in this code block

data = np.array(data)            # convert list of pairs into a n X 2 array
K_a_array = data[:,0] * 1E-4     # take the first item in each pair (first "column" of array)
k_acid_array = data[:,1] * 1E-3  # take the second item in each pair (second "column" of array)
# Note: The factors 1E-4 and 1E-3 come from the data table in the paper.

x = -np.log10(K_a_array)         # pKa
y = np.log10(k_acid_array)       # log k_obs


################################################################
################################################################
################################################################

# CODE BELOW IS MY RECOMMENDED PLOTTING METHOD COPIED AND REUSED HERE

## SETUP      ##################################################

# Install packages only if needed
# Coogle Colab does not include every Python package
#   
try:                  # if the package exists import it, otherwise install it
    import lmfit
except ImportError:
    !pip -q install lmfit
    import lmfit

try:
    import uncertainties
except ImportError:
    !pip -q install uncertainties
    import uncertainties

# Standard imports
from matplotlib import pyplot as plt
import numpy as np
import pandas as pd

import uncertainties as un
from uncertainties import unumpy as unp

from matplotlib.ticker import FormatStrFormatter
from scipy.stats import t

from pathlib import Path

Path("plots").mkdir(exist_ok=True)

try:
    import google.colab
    IN_COLAB = True
except ImportError:
    IN_COLAB = False

if IN_COLAB and not Path("tufte.mplstyle").exists():
    !wget -q https://raw.githubusercontent.com/blinkletter/Chem4410Webbook/main/book/styles/tufte.mplstyle

plt.rcdefaults()

print("Setup complete")


### LINE FIT    #################################################

# define the function for the model
# The model and parameters can be changed depening on if data_set == "acid" or not

if data_set == "acid":
    def bronsted(pKa, alpha, C):
        log_kobs = -alpha * pKa + C
        return log_kobs
    # Create a model by loading the function via the lmfit.Model tool
    model = lmfit.Model(bronsted)  

    # Set parameters - I can make a variable into a constant by setting vary=False.
    params = model.make_params(
        alpha = dict(
            value = 0.5,      # setting an initial guess. This is not necessary but can help in complex cases.
            vary = True,    # allow the parameter to vary in the optimization
        ),
        C = dict(       # 
            value = 0,      # set A_inf to zero
            vary = True,    # allow the parameter to vary in the optimization
        ), 
    )
else:
    def bronsted(pKa, beta, C):
        log_kobs = beta * pKa + C
        return log_kobs
    # Create a model by loading the function via the lmfit.Model tool
    model = lmfit.Model(bronsted)  

    # Set parameters - I can make a variable into a constant by setting vary=False.
    params = model.make_params(
        beta = dict(
            value = 0.5,      # setting an initial guess. This is not necessary but can help in complex cases.
            vary = True,    # allow the parameter to vary in the optimization
        ),
        C = dict(       # 
            value = 0,      # set A_inf to zero
            vary = True,    # allow the parameter to vary in the optimization
        ), 
    )

# use the .fit method on the model object to perform the curve fit
result = model.fit(y, params, pKa=x) 

print(result.fit_report())
print()
print(result.ci_report())

### BEST FIT LINE AND CONFIDENCE INTERVALS    ###################

# line fit
x_fit = np.linspace(np.min(x)-0.1, np.max(x)+0.1, 50)     # make an array of 50 points that span the data range
y_fit = result.eval(pKa=x_fit)                      # the line of the line fit

# confidence interval
y_err = result.eval_uncertainty(pKa=x_fit, sigma=2) # the 95% confidence interval
ci_upper = y_fit + y_err
ci_lower = y_fit - y_err

# Prediction interval

se = result.eval_uncertainty(pKa=x_fit, sigma=1)  # standard error of the fit
mse = result.redchi                             # reduced chi-squared value from the fit (mean squared error)
se_pred = np.sqrt(se**2 + mse)

dof = result.ndata - result.nvarys # degrees of freedom = number of data points - number of fitted parameters
tval = t.ppf(0.975,dof)            # two-tailed t-value for 95% confidence interval
pi = tval * se_pred

pi_lower = y_fit - pi
pi_upper = y_fit + pi

### PLOTTING    #################################################

# Fancy Plotting
# Note: We have x, y, x_fit, y_fit, ci_upper and ci_lower from previous calculations in the code blocks above.

from matplotlib.ticker import FormatStrFormatter

plt.rcdefaults()                 # reset plotting style
plt.style.use("tufte.mplstyle")  # apply the style sheet

fig, ax = plt.subplots(figsize = (4, 3.5))

ax.scatter(
    x, y, 
    marker = "o", s=32, 
    c="white", edgecolor = "black",
    linewidth = 0.7, 
    label=r"$\log{k_{obs}}$", 
    zorder=3
    )

ax.plot(
    x_fit, y_fit,
    color="black", 
    linewidth=0.5, zorder=2,
    )

ax.fill_between(
    x_fit, ci_lower, ci_upper,
    linewidth = 0,
    facecolor="gray", 
    edgecolor = "gray", 
    alpha=0.20,
    label="95% Confidence Band",
    zorder = 2,
    )

ax.spines["left"].set_position(("outward", 8))
ax.spines["bottom"].set_position(("outward", 8))
ax.set(
    ylabel=r"$\log{(k_{obs}\;/s^{-1})}$", 
    xlabel=r"$pK_a$",
#    xlim=[0,9], 
    xticks = x_ticks,                 
#    ylim=[-.2,3],     
    yticks = y_ticks,                 
      )
ax.yaxis.set_major_formatter(FormatStrFormatter('%.1f')) # 1 decimal place

fig.tight_layout()
fig.savefig("plots/result2.pdf")
ax.patch.set_facecolor([0, 0, 0, 0])          # Set face of plot to transparent
fig.savefig(f"plots/result2.png", dpi=600,    # save plot as .png with transparent background
            facecolor = [0, 0, 0, 0],
        )

plt.show()

```
